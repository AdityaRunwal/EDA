# Student Performance Analysis

## Exploratory Data Analysis (EDA)

### Objective
The objective of this project is to explore the student performance dataset, understand its structure, check data quality, and identify factors and patterns related to students' academic performance.

```python
# Importing the required libraries 

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## 1. Loading and Understanding the Dataset

```python
# Loading the dataset

df = pd.read_csv("student_data.csv")
df.head()
```

```python
# Checking the number of rows and columns in the dataset
df.shape
```

### Observation

The dataset contains 395 rows and 33 columns.
Each row represents a student, while each column represents a different feature related to the student's background, behavior, or academic performance.

```python
# Checking column names, data types, and non-null values
df.info()
```

### Observation

The dataset contains both numerical and categorical features. 
The info() function helps us understand the data type of each column and identify possible missing values.

```python
# Generating statistical summary of numerical columns
df.describe()
```

### Observation

The statistical summary provides an overview of the numerical features and helps identify their distribution, central tendency, spread, and possible unusual values.

```python
# Checking for missing values in each column
df.isnull().sum()
```

### Observation

No missing values were found in the dataset. Therefore, no missing value treatment is required.

```python
# Displaying all column names in the dataset
df.columns
```

```python
# Checking for duplicate records in the dataset
df.duplicated().sum()
```

### Observation

No duplicate records were found in the dataset. Therefore, no duplicate removal is required.

```python
# Checking the number of unique values in each column
df.nunique()
```

```python
# Checking the actual unique values in important categorical columns
categorical_cols = [
    'school',
    'sex',
    'address',
    'famsize',
    'Mjob',
    'Fjob',
    'higher',
    'internet',
    'romantic'
]

for col in categorical_cols:
    print(f"{col}: {df[col].unique()}")
```

```python
# Checking the minimum and maximum values of important numerical features
cols = ['age','studytime','failures','absences','G1','G2','G3']

df[cols].agg(['min','max'])
```

```python
# Visualizing the distribution of students' final grades

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='G3', bins=20, kde=True)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```
