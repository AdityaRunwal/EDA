# Student Performance Analysis

## Exploratory Data Analysis (EDA)

Exploring factors related to students' academic performance using the Student Performance dataset.

```python
# Importing required libraries

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Load Dataset

```python
# Loading the dataset

df = pd.read_csv("student_data.csv")

df.head()
```

```python
# Checking actual unique values in important categorical columns

categorical_cols = [
    'school',
    'sex',
    'address',
    'famsize',
    'Mjob',
    'Fjob',
    'higher',
    'internet',
    'romantic'
]

for col in categorical_cols:
    print(f"{col}: {df[col].unique()}")
```

## Exploratory Data Analysis

### Univariate Analysis

```python
# Distribution of final grades

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='G3', bins=20, kde=True)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```

### Gender Distribution

```python
# Number of students in each gender category

plt.figure(figsize=(6, 5))

sns.countplot(data=df, x='sex')

plt.title('Distribution of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Students')

plt.show()
```

### Study Time vs Final Grade

```python
# Comparing final grades across different study-time categories

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='studytime', y='G3')

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time Category')
plt.ylabel('Final Grade (G3)')

plt.show()
```

### Absences vs Final Grade

```python
# Examining the relationship between absences and final grades

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='absences', y='G3')

plt.title('Absences vs Final Grade')
plt.xlabel('Number of Absences')
plt.ylabel('Final Grade (G3)')

plt.show()
```

### Previous Failures vs Final Grade

```python
# Comparing final grades based on previous failures

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='failures', y='G3')

plt.title('Previous Failures vs Final Grade')
plt.xlabel('Number of Previous Failures')
plt.ylabel('Final Grade (G3)')

plt.show()
```

### Gender vs Final Grade

```python
# Comparing final grade distributions between genders

plt.figure(figsize=(6, 5))

sns.boxplot(data=df, x='sex', y='G3')

plt.title('Gender vs Final Grade')
plt.xlabel('Gender')
plt.ylabel('Final Grade (G3)')

plt.show()
```

## Correlation Analysis

```python
# Calculating correlations between numerical variables

correlation = df.corr(numeric_only=True)

correlation
```

```python
# Visualizing correlations using a heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')

plt.show()
```

## Conclusion

The exploratory analysis was performed to understand the distribution of student performance and examine relationships between academic performance and selected demographic, behavioral, and academic factors.

The analysis included:
- Distribution of final grades
- Gender distribution
- Study time and final grade
- Absences and final grade
- Previous failures and final grade
- Gender and final grade
